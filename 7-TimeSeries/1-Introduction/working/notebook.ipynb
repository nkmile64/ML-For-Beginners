The data to be analyzed, are taken from the **GEFCom2014 forecasting competition** and it consists of 3 years of *hourly electricity load* and *temperature* values between 2012 and 2014.

<h3 style="font-weight:600;color:white;background:#d53a49;padding:25px;margin:20;text-align:left;display:fill;border-radius:10px 0 0 10px;text-transform:uppercase;">Some Definitions:</h3>

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Time Series</span><br/>
A time series is a series of data points indexed in time order. The data points are usually taken at successive equally spaced points in time.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Time Series Analysis</span><br/>
The analysis of the data of a time series. For example, detection of patterns in the evolution of a time series, before and after an interrupting event.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Time Series Forecasting</span><br/>
Using a model to predict future values based on patterns observed on previously gathered data. The most common one is **ARIMA** (Autoregressive Integrated Moving Average) which works by relating the present value of a series to past values and past prediction errors. 

<h3 style="font-weight:600;color:white;background:#d53a49;padding:25px;margin:20;text-align:left;display:fill;border-radius:10px 0 0 10px;text-transform:uppercase;">Time Series Data Characteristics to consider</h3>

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Trends</span><br/>
Trends are measurable increases and decreases over time.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Seasonality</span><br/>
Seasonality is periodic fluctuations e.g. a holiday season.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Outliers</span><br/>
Points far away from the standard data variance.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Long-run cycle</span><br/>
Data may display a long-run cycle e.g. an economic downturn lasting more than a year.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Constant Variance</span><br/>
Data may display constant fluctuations over time e.g. energy usage per day and night.

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">Abrupt Changes</span><br/>
Changes due to unpredictable events e.g. the shuttering of businesses due to COVID.

<h3 style="font-weight:600;color:white;background:#d53a49;padding:25px;margin:20;text-align:left;display:fill;border-radius:10px 0 0 10px;text-transform:uppercase;">Power Usage Data</h3>

```python
import os
import matplotlib.pyplot as plt
from common.utils import load_data
```

Examine the data as a dataframe calling `load_data()` and `head()`. Note that the first method returns a dataframe with a DateTimeIndex and two columns: load and temp. Below we "create" a new dataframe with only the `load` column:

```python
data_dir = "./data"
energy = load_data(data_dir)[["load"]]
energy.head()
```

Plot the data using `plot()`:

```python
energy.plot(y="load", subplots=True, figsize=(18, 8), fontsize=12)
plt.xlabel("timestamp", fontsize=12)
plt.ylabel("load", fontsize=12)

plt.show()
```

Next, plot the first week of July 2014 - use the pattern: from data : to date.

```python
energy["2014-07-01":"2014-07-07"].plot(y="load", subplots=True, figsize=(15, 8), fontsize=12)
plt.xlabel("timestamp", fontsize=12)
plt.ylabel("load", fontsize=12)
plt.title("First week of July 2014", fontsize=20)

plt.show()
```

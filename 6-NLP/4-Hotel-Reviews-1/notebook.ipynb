<h1 style="color:#d53a49;font-weight:600;">Sentiment Analysis with Hotel Reviews</h1>

> **OBJECTIVE:** Build a hotel recommendation bot using *sentiment analysis* and guest *reviews scores*
>
> This dataset contains 515,000 customer reviews and scoring of 1493 luxury hotels across Europe. Meanwhile, the geographical location of hotels are also provided for further analysis.
>
> The csv file contains 17 fields. The description of each field is as below:

---
- **Hotel_Address:** Address of hotel.
- **Review_Date:** Date when reviewer posted the corresponding review.
- **Average_Score:** Average Score of the hotel, calculated based on the latest comment in the last year.
- **Hotel_Name:** Name of Hotel
- **Reviewer_Nationality:** Nationality of Reviewer
- **Negative_Review:** Negative Review the reviewer gave to the hotel. If the reviewer wrote nothing, then it should be: 'No Negative'
- **Review_Total_Negative_Word_Counts:** Total number of words in the negative review.
- **Positive_Review:** Positive Review the reviewer gave to the hotel. If the reviewer wrote nothing, then it should be: 'No Positive'
- **Review_Total_Positive_Word_Counts:** Total number of words in the positive review.
- **Reviewer_Score:** Score the reviewer has given to the hotel, based on his/her experience. The values range between 2.5 - 10.
- **Total_Number_of_Reviews_Reviewer_Has_Given:** Number of Reviews the reviewers has given in the past.
- **Total_Number_of_Reviews:** Total number of valid reviews the hotel has.
- **Tags:** Tags reviewer gave the hotel.
- **days_since_review:** Duration between the review date and scrape date. Note that older reviews might not be as accurate as newer ones due to changes.
- **Additional_Number_of_Scoring:** There are also some guests who just made a scoring on the service rather than a review. This number indicates how many valid scores without review in there.
- **lat:** Latitude of the hotel
- **lng:** longtitude of the hotel

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">2. Calculate the frequency count for reviewer nationalities</span>

<span>i. How many distinct values are there for the column`Reviewer_Nationality` and what are they?</span><br/>

Use `value_counts()` which returns a Series that has as **index** the country and as **values** the frequency they occur.

```python
nationality_freq = df["Reviewer_Nationality"].value_counts()
num_of_nationalities = nationality_freq.size
print(f"There are {num_of_nationalities} different nationalities.", end="\n\n")
print(nationality_freq.to_string())
```

<span>ii. What reviewer nationality is the most common in the dataset (print country and number of reviews)</span><br/>
<span>iii. What are the next top-10 most frequently found nationalities, and their frequency count?</span>

```python
top_nation = nationality_freq.index[0].strip()
top_nation_freq = nationality_freq.values[0]
next_ten_nation = nationality_freq[1:11].to_string()
print(f'The highest-frequency reviewer nationality is "{top_nation}" with {top_nation_freq} reviews.')
print(f"The next 10 higher-frequency reviewer nationalities are:\n\n{next_ten_nation}")
```

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">3. What was the most frequently reviewed hotel for each of the top-10 reviewer nationalities</span>

Start with a loop and extract all the rows that match the criteria (a specific nationality) into a new dataframe. Then for each country get the hotel frequency.

```python
for nat in nationality_freq.index[:10]:
    nat_df = df[df["Reviewer_Nationality"] == nat]
    freq = nat_df["Hotel_Name"].value_counts()
    print(f"The most reviewed hotel for {nat.strip()} was {freq.index[0]} with {freq.values[0]} reviews.")
```

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">4. How many reviews are there per hotel in the dataset?</span>

We start by creating a new DataFrame from the old one, with only the required columns. Then, we group the rows by "Hotel_Name", count them and put the result in a new column "Total_Reviews_Found". Finally, get rid of all the duplicated rows.

**IMPORTANT NOTE**: We don't use count() because this is an aggregate function producing one row per hotel. Instead, transform("count") produces the same summary but broadcasts the result back to the original rows.

```python
# fmt:off
hotel_freq_df = df[["Hotel_Name", "Total_Number_of_Reviews"]]
hotel_freq_df["Total_Reviews_Found"] = (
    hotel_freq_df.groupby("Hotel_Name")
                 .transform("count")
)
hotel_freq_df = (
    hotel_freq_df.drop_duplicates(subset=["Hotel_Name"])
                 .sort_values(by="Total_Reviews_Found", ascending=False)
)
display(hotel_freq_df)
```

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">5. Calculate an average review score for each hotel</span>

While there is an `Average_Score` column for each hotel in the dataset, we can also calculate an average score - the average of all reviewer scores for each hotel.<br/>
A such, we add a new column to the dataframe with the header `Calc_Average_Score` that contains that calculated average. Print out the columns `Hotel_Name`, `Average_Score`, and `Calc_Average_score`.

```python
# fmt: off
def get_diff_review_avg(row):
    """
    Takes a row and returns the difference between average score and calculated average score.
    """
    return row["Average_Score"] - row["Calc_Average_Score"]


df["Calc_Average_Score"] = (
    round(df.groupby("Hotel_Name")["Reviewer_Score"]
            .transform("mean"),1)
)

df["Average_Score_Difference"] = df.apply(get_diff_review_avg, axis=1)

review_scores_df = (
    df.drop_duplicates(subset=["Hotel_Name"])
      .sort_values(["Average_Score_Difference"])
)

display(review_scores_df[["Average_Score_Difference", "Average_Score", "Calc_Average_Score", "Hotel_Name"]])
```

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">6. Calculate and print out how many rows have column `Negative_Review` values of "No Negative"</span>

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">7. Calculate and print out how many rows have column `Positive_Review` values of "No Positive"</span>

<span style="color:#d53a49;text-transform:uppercase;font-weight:600;">8. Calculate and print out how many rows have column `Positive_Review` values of "No Positive" **and** `Negative_Review` values of "No Negative"</span>

```python
start = time.time()
no_negative_reviews = df.apply(lambda x: True if x["Negative_Review"] == "No Negative" else False, axis=1)
print(f"Number of No Negative reviews: {len(no_negative_reviews[no_negative_reviews==True].index)}")

no_positive_reviews = df.apply(lambda x: True if x["Positive_Review"] == "No Positive" else False, axis=1)
print(f"Number of No Positive reviews: {len(no_positive_reviews[no_positive_reviews==True].index)}")

both_no_reviews = df.apply(
    lambda x: True if x["Negative_Review"] == "No Negative" and x["Positive_Review"] == "No Positive" else False,
    axis=1,
)
print(f"Number of both No Negative and No Positive reviews: {len(both_no_reviews[both_no_reviews==True].index)}")

end = time.time()
print(f"Lambdas took {round(end-start,2)} seconds.")
```

Using lambdas seems to take a long time! Let's try and use `sum` to count the rows:

```python
start = time.time()

no_negative_reviews = sum(df.Negative_Review == "No Negative")
print(f"Number of negative reviews: {no_negative_reviews}")

no_positive_reviews = sum(df.Positive_Review == "No Positive")
print(f"Number of positive reviews: {no_positive_reviews}")

both_no_reviews = sum((df.Negative_Review == "No Negative") & (df.Positive_Review == "No Positive"))
print(f"Number of both No Negative and No Positive reviews: {both_no_reviews}")

end = time.time()
print(f"Sum took {round(end-start,2)} seconds.")
```

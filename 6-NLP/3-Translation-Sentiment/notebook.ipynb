### Translation & Sentiment Analysis with ML

#### The Naive Approach

A naive translation would involve two steps (ignoring sentence structure):
- **Identification** : tag the input worlds (noun, verb, etc)
- **Create translation** : directly translate each word

#### Machine Learning Approaches

Let's try the first line from *Pride and Prejudice*.

```python
from textblob import TextBlob
from googletrans import Translator

blob = TextBlob(
    "It is a truth universally acknowledged, that a single man in  possession of a good fortune, must be in want of a wife!"
)

# Use googletrans for translation
translator = Translator()
translated = translator.translate(blob.raw, src="en", dest="el")

print(translated.text)
```

> NOTE: the above translation is actually done by googletrans so we could just pass the string to the translator (without instantiating the TextBlob)

### Sentiment Analysis

#### ML Approaches

Sentiment is measured with a **polarity**, where:
- -1 is the most negative sentiment
- +1 is the most positive

Sentiment is also measured with a **subjectivity** where:
- 0 is the most objective
- 1 is the most subjective

Let's analyze first and last sentences from *Pride and Prejudice*

```python
from textblob import TextBlob

quote1 = """It is a truth universally acknowledged, that a single man in possession of a good fortune, must be in want of a wife."""

quote2 = """Darcy, as well as Elizabeth, really loved them; and they were both ever sensible of the warmest gratitude towards the persons who, by bringing her into Derbyshire, had been the means of uniting them."""

sentiment1 = TextBlob(quote1).sentiment
sentiment2 = TextBlob(quote2).sentiment

print(f"'{quote1}' has a sentiment of:\n{sentiment1}\n")
print(f"'{quote2}' has a sentiment of:\n{sentiment2}")
```

Next, we'll use sentiment polarity to check if *Pride and Prejudice* has more absolutely positive sentences than absolutely negative sentences:

```python
from textblob import TextBlob

# Open the file and extract the contents as a string:
path = "../data/Jane Austin.txt"

with open(file=path, mode="r", encoding="utf-8") as file:
    file = file.read()

blob = TextBlob(file)
positive = []
negative = []

# Analyze each sentence in the book, in a loop:
for sentence in blob.sentences:
    polarity = sentence.sentiment.polarity
    if polarity == 1:
        positive.append(sentence)
    elif polarity == -1:
        negative.append(sentence)

print(f"positive sentences : {len(positive)}")
for sentence in positive:
    print(f" + {sentence.replace('\n','')}")

print(f"negative sentences : {len(negative)}")
for sentence in negative:
    print(f" - {sentence.replace('\n','')}")
```
